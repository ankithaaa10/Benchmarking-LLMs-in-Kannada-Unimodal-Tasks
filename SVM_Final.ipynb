```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
from sklearn.metrics import classification_report, accuracy_score
from sklearn.model_selection import GridSearchCV
```

```python
# Function to load data from text files
def load_data(file_path):
    data = []
    labels = []
    with open(file_path, 'r', encoding='utf-8') as f:
        for line in f:
            # Split at the last comma (handles text with commas)
            text, label = line.strip().rsplit(',', 1)
            data.append(text.strip())
            labels.append(label.strip())
    return data, labels
```

```python
# --- STEP 1: Load the data ---
# Replace with your actual file paths if different
train_file = '/content/train .txt'  # Format: "text,label" per line
test_file = '/content/test.txt'    # Format: "text,label" per line
```

```python
#Load training and testing data
X_train, y_train = load_data(train_file)  # Texts & labels for training
X_test, y_test = load_data(test_file)     # Texts & labels for testing
```

```python
# --- STEP 2: Convert text to numerical features (TF-IDF) ---
vectorizer = TfidfVectorizer()
X_train_tfidf = vectorizer.fit_transform(X_train)  # Train TF-IDF
X_test_tfidf = vectorizer.transform(X_test)        # Apply to test data
```

```python
# --- STEP 3: Train SVM with hyperparameter tuning ---
param_grid = {
    'C': [0.1, 1, 10, 100],      # Regularization strength
    'kernel': ['linear', 'rbf'],  # Kernel type
    'gamma': ['scale', 'auto']    # Kernel coefficient
}
```

```python
svm = SVC()
grid_search = GridSearchCV(svm, param_grid, cv=5, scoring='accuracy', n_jobs=-1)
grid_search.fit(X_train_tfidf, y_train)  # Find best hyperparameters
```

```python
# --- STEP 4: Evaluate the best model ---
best_svm = grid_search.best_estimator_
y_pred = best_svm.predict(X_test_tfidf)
```

```python
print("Best Parameters:", grid_search.best_params_)
print("\nAccuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```
