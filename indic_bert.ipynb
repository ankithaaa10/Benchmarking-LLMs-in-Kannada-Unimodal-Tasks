```python
import pandas as pd

def load_data(filepath):
    texts = []
    labels = []
    with open(filepath, 'r', encoding='utf-8') as file:
        for line in file:
            if not line.strip():
                continue  # skip empty lines
            try:
                text, label = line.strip().rsplit(',', 1)
                texts.append(text)
                labels.append(label)
            except ValueError:
                print(f"Skipped line: {line.strip()}")
    return pd.DataFrame({'text': texts, 'label': labels})  # ✅ returns DataFrame
```

```python
df_train = load_data("/content/train .txt")
df_test = load_data("/content/test.txt")
```

```python
# Now this will work without error
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()
df_train['label'] = label_encoder.fit_transform(df_train['label'])
df_test['label'] = label_encoder.transform(df_test['label'])

label2id = {label: int(idx) for idx, label in enumerate(df_train['label'].unique())}
id2label = {v: k for k, v in label2id.items()}
```

```python
from transformers import AutoTokenizer

model_name = "ai4bharat/indic-bert"
tokenizer = AutoTokenizer.from_pretrained(model_name)

def tokenize_function(example):
    return tokenizer(example["text"], padding="max_length", truncation=True, max_length=128)

from datasets import Dataset

train_dataset = Dataset.from_pandas(df_train)
test_dataset = Dataset.from_pandas(df_test)

train_dataset = train_dataset.map(tokenize_function, batched=True)
test_dataset = test_dataset.map(tokenize_function, batched=True)

train_dataset = train_dataset.rename_column("label", "labels")
test_dataset = test_dataset.rename_column("label", "labels")

train_dataset.set_format("torch", columns=["input_ids", "attention_mask", "labels"])
test_dataset.set_format("torch", columns=["input_ids", "attention_mask", "labels"])
```

```python
# Fix: Ensure all keys and values are Python native types
import numpy as np

unique_labels = df_train['label'].unique()
label2id = {str(label): int(i) for i, label in enumerate(unique_labels)}
id2label = {int(i): str(label) for label, i in label2id.items()}
```

```python
print(label2id)
print(id2label)
print(type(list(label2id.values())[0]))  # should print: <class 'int'>
```

```python
from transformers import AutoModelForSequenceClassification

model = AutoModelForSequenceClassification.from_pretrained(
    model_name,
    num_labels=len(label2id),
    id2label=id2label,
    label2id=label2id
)

```

```python
from transformers import TrainingArguments, Trainer
from sklearn.metrics import accuracy_score, classification_report

def compute_metrics(eval_pred):
    logits, labels = eval_pred
    preds = logits.argmax(axis=1)
    acc = accuracy_score(labels, preds)
    return {"accuracy": acc}

training_args = TrainingArguments(
    output_dir="./indicbert-kannada-classifier",
    eval_strategy="epoch",
    save_strategy="epoch",
    learning_rate=2e-5,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,
    num_train_epochs=5,
    weight_decay=0.01,
    logging_dir="./logs",
    load_best_model_at_end=True,
    metric_for_best_model="accuracy"
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=test_dataset,
    compute_metrics=compute_metrics
)
```

```python
trainer.train()
```

```python
trainer.evaluate()

# Optional: Get predictions
predictions = trainer.predict(test_dataset)
pred_labels = predictions.predictions.argmax(axis=1)

# Decode predicted and true labels
true_labels = df_test['label']
decoded_preds = label_encoder.inverse_transform(pred_labels)
decoded_trues = label_encoder.inverse_transform(true_labels)

print(classification_report(decoded_trues, decoded_preds))
```
